# 07 Classification Metrics

```python
# (a)
tp = 45 # shown
fp = 15 # 60 - 45
tn = 165 # 190 - 25
fn = 25 # shown

# (b)
acc = (tp + tn) / (tp + tn + fp + fn)
print('accuracy=', acc)

pre = (tp) / (tp + fp)
print('precision=', pre)

re = (tp) / (tp + fn)
print('recall=', re)

f1 = 2 * ((pre * re) / (pre + re))
print('F1=', f1)
```
